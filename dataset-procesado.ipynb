## Intro

Prepara tus fotos originales para que puedan ser usadas por un modelo de aprendizaje automático. Eso implica:

- Redimensionar las imagenes a un tamaño estándar (224x224 píxeles).
- Normalizar las imagenes (escalar los valores de los píxeles entre 0 y 1).
- Guardar las imagenes organizadas para entrenamiento posterior.

```python
### Importaciones necesarios

import os
import cv2
import numpy as np
```

```python
### Variables

data_dir = "dataset"
output_dir = "dataset_procesado"
img_size = (224, 224)
```

```python
if not os.path.exists(output_dir):
    os.makedirs(output_dir)

### Recorro el directorio de datos y creo un nuevo directorio para cada clase
for subdir, dirs, files in os.walk(data_dir):
        for file in files:
            if file.lower().endswith(('.png', '.jpg', '.jpeg')):
                filepath = os.path.join(subdir, file)
                image = cv2.imread(filepath)

                if image is None:
                    print(f"⚠️ Error al leer la imagen: {filepath}")
                    continue

                # Redimensionar
                resized_image = cv2.resize(image, img_size)

                # Normalizar (valores entre 0 y 1)
                normalized_image = resized_image / 255.0

                # Crear estructura de carpetas
                relative_path = os.path.relpath(subdir, data_dir)
                output_subdir = os.path.join(output_dir, relative_path)
                os.makedirs(output_subdir, exist_ok=True)

                # Guardar imagen procesada
                output_filepath = os.path.join(output_subdir, file)
                cv2.imwrite(output_filepath, (normalized_image * 255).astype(np.uint8))
                print(f"✅ Imagen guardada en: {output_filepath}")
```
