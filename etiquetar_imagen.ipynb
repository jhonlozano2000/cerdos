```python
import cv2
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.image import img_to_array
import os
from datetime import datetime

# Configuración
model_path = 'modelo_identificacion_cerdos.h5'
dataset_dir = 'dataset_procesado'
img_size = (224, 224)
imagen_a_analizar = '20250324_121238.jpg'  # Imagen que quieres probar

# Cargar modelo
model = tf.keras.models.load_model(model_path)

# Obtener clases desde las carpetas del dataset procesado
class_names = sorted(os.listdir(dataset_dir))
class_indices = {i: name for i, name in enumerate(class_names)}

# Preprocesamiento
def preprocess_image(image_path):
    image = cv2.imread(image_path)
    if image is None:
        raise ValueError(f"No se pudo cargar la imagen: {image_path}")
    image = cv2.resize(image, img_size)
    img_array = img_to_array(image)
    img_array = np.expand_dims(img_array, axis=0)
    img_array = img_array / 255.0
    return img_array

# Predicción
def etiquetar_imagen(image_path):
    img = preprocess_image(image_path)
    prediction = model.predict(img)
    class_index = np.argmax(prediction)
    confidence = np.max(prediction)
    predicted_name = class_indices[class_index]

    # Resultado
    print(f"\n📷 Imagen: {image_path}")
    print(f"🐷 Predicción: {predicted_name}")
    print(f"📊 Confianza: {confidence:.2f}")

    # Registrar en archivo
    with open('registro_etiquetas.txt', 'a') as log:
        log.write(f"{datetime.now()} - {image_path} -> {predicted_name} ({confidence:.2f})\n")

# Ejecutar
etiquetar_imagen(imagen_a_analizar)
```
